# reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes the input tensor to a defined shape
* View - return the view tensor of the input tensor
* stacking
* Squeezing
* Unsqueezing
* Permute



```python
# create a tensor
x = torch.arange(1., 10.)
x, x.shape
```

```python
# reshaping a tensor  # addinng Dimensions to the shapes and tensors
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# add 2 dimenisons
x_reshaped = x.reshape(9, 1)
x_reshaped, x_reshaped.shape
```

```python
# add 2 dimenisions
x_reshaped = x.reshape(3, 3)
x_reshaped, x_reshaped.shape
```

```python
# change the View of the tensors
z = x.view(1,9)
z, z.shape
```

```python
# changing  CHanges the x (view of a tensor shares the same memory as the orginal one)
```

```python
# stacking the tensors on each other

x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# using Squeeze in the target tensor
x_reshaped, x_reshaped.shape
```

# Pytorch and Numpy


* Data in numpy array, want them in Pytorch Tensor -> torch.from_numpy(ndarry)
* Pytorch Tensor -> Numpy =-> torch.numpy()


```python
import numpy as np
```

```python
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array).type(torch.float32) # by defualt, it brings out the default float64, so you need to make it float32
array, tensor
```

```python
array = array + 1
```

```python
array,  tensor
```

```python
# Tensor to Numpy array
tensor = torch.ones(7)

numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
# change the tensor to the arry
tensor = tensor + 1
tensor, numpy_tensor
```


## Reproducbility (trying to take random out of Random)

`start with random numbers -> tensor operations -> update random numbers to try to make she operations of the date -> again -> again ...

To reduce the randomnese in neural networks and pytorch comes the concept of the random seed



```python
torch.rand(3,3)
```

```python
torch.rand(3,3)
```

```python
# create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
